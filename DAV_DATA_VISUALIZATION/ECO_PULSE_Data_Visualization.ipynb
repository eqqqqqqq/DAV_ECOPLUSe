```python
import pandas as pd
import numpy as np
import plotly.express as px

# Load the ECO PULSE dataset
df = pd.read_excel("ECO_PULSE_Cleaned_Data.xlsx")

# Display the first 5 rows
df.head()
```

```python
# Display the number of rows and columns
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display dataset information
df.info()

# Statistical summary
df.describe()
```

```python
# 1. Bar Chart — Dengue Cases by State

cases_by_state = df.groupby("State", as_index=False)["Dengue_Cases_2024"].sum()

fig = px.bar(
    cases_by_state,
    x="State",
    y="Dengue_Cases_2024",
    title="Total Dengue Cases in 2024 by State",
    labels={"State": "State", "Dengue_Cases_2024": "Dengue Cases"},
    text_auto=".2s"
)
fig.show()
```

```python
# 2. Pie Chart — Distribution of Dengue Cases by Urban/Rural Area

area_cases = df.groupby("Urban_Rural", as_index=False)["Dengue_Cases_2024"].sum()

fig = px.pie(
    area_cases,
    names="Urban_Rural",
    values="Dengue_Cases_2024",
    title="Dengue Case Distribution by Area",
    hole=0.3
)
fig.show()
```

```python
# 3. Histogram — Age Distribution

fig = px.histogram(
    df,
    x="Age",
    nbins=20,
    title="Age Distribution of People",
    labels={"Age": "Age"}
)
fig.show()
```

```python
# 4. Scatter Plot — Temperature vs Dengue Cases

fig = px.scatter(
    df,
    x="Avg_Temp_Max_C",
    y="Dengue_Cases_2024",
    color="State",
    title="Maximum Temperature vs Dengue Cases",
    labels={
        "Avg_Temp_Max_C": "Average Maximum Temperature (°C)",
        "Dengue_Cases_2024": "Dengue Cases"
    },
    hover_data=["Station_City", "State"]
)
fig.show()
```

```python
# 5. Correlation Heatmap — Numerical Variables

numeric_df = df.select_dtypes(include=["number"])
correlation = numeric_df.corr()

fig = px.imshow(
    correlation,
    text_auto=".2f",
    aspect="auto",
    title="Correlation Heatmap of ECO PULSE Numerical Variables"
)
fig.show()
```

```python
# 6. Bar Chart — Average Environmental Dengue Risk by State

risk_by_state = df.groupby("State", as_index=False)["Environmental_Dengue_Risk_Index"].mean()
risk_by_state["Environmental_Dengue_Risk_Index"] = risk_by_state["Environmental_Dengue_Risk_Index"].round(2)

fig = px.bar(
    risk_by_state,
    x="State",
    y="Environmental_Dengue_Risk_Index",
    title="Average Environmental Dengue Risk Index by State",
    labels={"State": "State", "Environmental_Dengue_Risk_Index": "Average Risk Index"},
    text_auto=True
)
fig.show()
```

```python
# 7. Bar Chart — Dengue Deaths by State

deaths_by_state = df.groupby("State", as_index=False)["Dengue_Deaths_2019_2024"].sum()

fig = px.bar(
    deaths_by_state,
    x="State",
    y="Dengue_Deaths_2019_2024",
    title="Dengue Deaths (2019–2024) by State",
    labels={"State": "State", "Dengue_Deaths_2019_2024": "Dengue Deaths"},
    text_auto=".2s"
)
fig.show()
```

```python
# 8. Pie Chart — Records by Gender

gender_counts = df["Gender"].value_counts().reset_index()
gender_counts.columns = ["Gender", "Number_of_People"]

fig = px.pie(
    gender_counts,
    names="Gender",
    values="Number_of_People",
    title="Distribution of People by Gender"
)
fig.show()
```

```python
# 9. Bar Chart — Average Hospital Rating by State

hospital_rating = df.groupby("State", as_index=False)["Avg_Hospital_Rating"].mean()
hospital_rating["Avg_Hospital_Rating"] = hospital_rating["Avg_Hospital_Rating"].round(2)

fig = px.bar(
    hospital_rating,
    x="State",
    y="Avg_Hospital_Rating",
    title="Average Hospital Rating by State",
    labels={"State": "State", "Avg_Hospital_Rating": "Average Hospital Rating"},
    text_auto=True
)
fig.show()
```

```python
# 10. Histogram — Environmental Dengue Risk Index Distribution

fig = px.histogram(
    df,
    x="Environmental_Dengue_Risk_Index",
    nbins=20,
    title="Environmental Dengue Risk Index Distribution",
    labels={"Environmental_Dengue_Risk_Index": "Environmental Dengue Risk Index"}
)
fig.show()
```

```python
# 11. Box Plot — Dengue Cases by Urban/Rural Area

fig = px.box(
    df,
    x="Urban_Rural",
    y="Dengue_Cases_2024",
    color="Urban_Rural",
    title="Dengue Cases Distribution by Urban/Rural Area",
    labels={"Urban_Rural": "Area Type", "Dengue_Cases_2024": "Dengue Cases in 2024"}
)
fig.show()
```

```python
# 12. Bar Chart — Top 10 Cities by Dengue Cases

cases_by_city = df.groupby("Station_City", as_index=False)["Dengue_Cases_2024"].sum()
top_10_cities = cases_by_city.nlargest(10, "Dengue_Cases_2024")

fig = px.bar(
    top_10_cities,
    x="Station_City",
    y="Dengue_Cases_2024",
    title="Top 10 Cities by Dengue Cases in 2024",
    labels={"Station_City": "City", "Dengue_Cases_2024": "Dengue Cases"},
    text_auto=".2s"
)
fig.show()
```

```python
# 13. Bar Chart — Average Rainfall by State

rain_by_state = df.groupby("State", as_index=False)["Avg_Rain_mm"].mean()
rain_by_state["Avg_Rain_mm"] = rain_by_state["Avg_Rain_mm"].round(2)

fig = px.bar(
    rain_by_state,
    x="State",
    y="Avg_Rain_mm",
    title="Average Rainfall by State",
    labels={"State": "State", "Avg_Rain_mm": "Average Rainfall (mm)"},
    text_auto=True
)
fig.show()
```

```python
# 14. Scatter Plot — Rainfall vs Environmental Dengue Risk

fig = px.scatter(
    df,
    x="Avg_Rain_mm",
    y="Environmental_Dengue_Risk_Index",
    color="Urban_Rural",
    title="Rainfall vs Environmental Dengue Risk Index",
    labels={
        "Avg_Rain_mm": "Average Rainfall (mm)",
        "Environmental_Dengue_Risk_Index": "Environmental Dengue Risk Index"
    },
    hover_data=["Station_City", "State"]
)
fig.show()
```

```python
# 15. Bar Chart — Total Hospital Count by State

hospitals_by_state = df.groupby("State", as_index=False)["Hospital_Count"].sum()

fig = px.bar(
    hospitals_by_state,
    x="State",
    y="Hospital_Count",
    title="Total Hospital Count by State",
    labels={"State": "State", "Hospital_Count": "Hospital Count"},
    text_auto=".2s"
)
fig.show()
```

```python
# 16. Geographical Map — Environmental Dengue Risk by Location

# Remove duplicate coordinates so each location appears once
location_risk = df.drop_duplicates(subset=["Station_City", "latitude", "longitude"])

fig = px.scatter_geo(
    location_risk,
    lat="latitude",
    lon="longitude",
    color="Environmental_Dengue_Risk_Index",
    size="Dengue_Cases_2024",
    hover_name="Station_City",
    hover_data=["State", "Dengue_Cases_2024", "Environmental_Dengue_Risk_Index"],
    scope="asia",
    title="Environmental Dengue Risk Across Indian Locations"
)
fig.show()
```
